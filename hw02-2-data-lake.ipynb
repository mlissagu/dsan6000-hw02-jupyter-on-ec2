```python
import boto3

s3 = boto3.client("s3")
bucket_name = "dsan6000-ml2275"
s3.create_bucket(
    Bucket=bucket_name
)
```

```python
from pathlib import Path

data_dir = Path("data")

for file_path in data_dir.glob("*.parquet"):
    s3_key = f"wikipedia-hourly/{file_path.name}"
    
    s3.upload_file(
        str(file_path),
        bucket_name,
        s3_key
    )

print("Upload complete")
```

```python
response = s3.list_objects_v2(
    Bucket=bucket_name,
    Prefix="wikipedia-hourly/"
)

for obj in response.get("Contents", []):
    print(obj["Key"])
```
