```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

bucket = "dsan6000-ml2275"
files = [
    f"s3://{bucket}/wikipedia-hourly/{f.name}"
    for f in Path("data").glob("*.parquet")
]

dfs = [pd.read_parquet(uri) for uri in files]
df = pd.concat(dfs, ignore_index=True)

df.head()
```

```python
Path("images").mkdir(exist_ok=True)

hourly = df.groupby("hour").size().reset_index(name="events")

sns.lineplot(data=hourly, x="hour", y="events", marker="o")
plt.title("Wikipedia Events per Hour")
plt.tight_layout()
plt.savefig("images/hourly-events.png")
plt.savefig("images/hourly-events.svg")
plt.show()
```

```python
hourly_type = (
    df.groupby(["hour", "type"])
      .size()
      .reset_index(name="events")
)

sns.lineplot(
    data=hourly_type,
    x="hour",
    y="events",
    hue="type",
    marker="o"
)
plt.title("Wikipedia Events per Hour by Type")
plt.tight_layout()
plt.savefig("images/events-by-type.png")
plt.savefig("images/events-by-type.svg")
plt.show()
```
